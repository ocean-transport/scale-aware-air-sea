```python
# set up a super simple example of long-running task
import time

@dask.delayed
def long(x):
    # time.sleep(4*60) # simulate our filtering/air-sea calc
    time.sleep(5*60)
    return x + 1
```

```python
data = range(4)
output = []
for x in data:
    a = long(x)
    output.append(a)
total = dask.delayed(sum)(output)
```
